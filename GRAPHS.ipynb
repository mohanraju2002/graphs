```python
# LINE GRAPH

import matplotlib.pyplot as plt
days=["mon","tues","wed","thur","fri"]
earn=[300,450,200,400,650]
plt.plot(days,earn,marker="*",label="earn")
plt.xlabel("days")
plt.ylabel("earning")
plt.title(" LINE GRAPH",color="red")
plt.grid(True)
plt.legend()
plt.show()

```

```python
# LINE GRAPH WITH 2 LINES

import matplotlib.pyplot as plt
day=["mon","tues","wed","thurs","fri"]
drinks=[300,450,150,400,650]
food=[400,500,350,300,500]
plt.plot(day,drinks,marker="*",color="blue",label="drinksSS")
plt.plot(day,food,marker="o",color="red",label="food")
plt.xlabel("days")
plt.ylabel("items")
plt.title("LINE GRAPH WITH 2 LINES",color="red")
plt.grid(True)
plt.legend()
plt.show()
```

```python
# BAR GRAPH (VERICAL)
import matplotlib.pyplot as plt
months=["january","february","march"]
visitors=[150,300,250]
plt.bar(months,visitors,color=("red","blue","green"))
plt.xlabel("months")
plt.ylabel("no of visitors")
plt.title("vertical bar chart",color="red")
plt.show()
```

```python
# HORZENTAL
import matplotlib.pyplot as plt
month=["jan","feb","mar"]
visitors=[150,300,250]
plt.barh(month,visitors,color=("red","blue","green"))
plt.xlabel("number of visitors")
plt.ylabel("months")
plt.title("HORZONTAL BAR CHART",color="red")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
month=["april","may","june"]
adult=[300,500,700]
child=[200,600,600]
x=np.arange(len(month))
width=0.25
plt.bar(x-width,adult,width,label="adults",color="red")
plt.bar(x,child,width,label="child",color="blue")
plt.xlabel("months")
plt.ylabel("visitors")
plt.title("2 CATEGORIES",color="red")
plt.legend()
plt.xticks(x,month)
plt.show()

```

```python
import numpy as np
import matplotlib.pyplot as plt
month=["april","may","june"]
adult=[300,500,700]
child=[245,600,200]
mens=[100,300,400]
x=np.arange(len(month))
width=0.25
plt.bar(x-width,adult,width,label="adults",color="red")
plt.bar(x,child,width,label="child",color="blue")
plt.bar(x+width,mens,width,label="mens",color="green")
plt.xlabel("months")
plt.ylabel("visitors")
plt.title("3 CATEGORIES",color="red")
plt.legend()
plt.xticks(x,month)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
month=["april","may","june"]
adult=[300,500,700]
child=[245,600,200]
mens=[100,300,400]
womens=[457,678,444]
x=np.arange(len(month))
width=0.18
plt.bar(x-1.5*width,adult,width,label="adults",color="red")
plt.bar(x-0.5*width,child,width,label="child",color="blue")
plt.bar(x+0.5*width,mens,width,label="mens",color="green")
plt.bar(x+1.5*width,womens,width,label="womens",color="pink")
plt.xlabel("months")
plt.ylabel("visitors")
plt.title("BY USING 4 CATEGORIES ",color="red")
plt.legend()
plt.xticks(x,month)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
month=["april","may","june"]
adult=[300,500,700]
child=[245,600,200]
mens=[100,300,400]
womens=[457,678,444]
old=[345,600,666]
x=np.arange(len(month))
width=0.15
plt.bar(x-2*width,adult,width,label="adults",color="red")
plt.bar(x-width,child,width,label="child",color="blue")
plt.bar(x,mens,width,label="mens",color="green")
plt.bar(x+width,womens,width,label="womens",color="pink")
plt.bar(x+2*width,old,width,label="old",color="black")
plt.xlabel("months")
plt.ylabel("visitors")
plt.title("BY USING 5 CATEGORIES ",color="red")
plt.legend()
plt.xticks(x,month)
plt.show()


```

```python
import matplotlib.pyplot as plt
movies=["comedy","action","romance","scifi","drama"]
views=[4,5,6,4,1]
plt.pie(views,labels=movies,autopct="%1.1f%%")
plt.title("PIE CHART OF FAVORITE MOVIE")
plt.show()
```

```python
# SCATTER PLOT
import matplotlib.pyplot as plt
age=[18,25,57,45,26,64,37,40,24,33]
income=[15000,29000,68000,52000,32000,80000,41000,45000,26000,33000]
plt.scatter(age,income,marker="d")
plt.xlabel("age")
plt.ylabel("yearly income")
plt.title("income")
plt.grid(True)
plt.show()
```
