# ***LABEL_ENCODER***

```python
from sklearn.preprocessing import LabelEncoder
import pandas as pd

data={
    "player":["Abd","Virat","rohith","gg"],
    "team":["rcb","rcb","mi","kkr"]
}

df=pd.DataFrame(data)
labelencoder=LabelEncoder()

df["player_n"]=labelencoder.fit_transform(df["player"])
df["team_n"]=labelencoder.fit_transform(df["team"])
print(df)
```

# ***MIN_MAX SCALER***

# ***HANDLING_OUTLIERS WITH WHERE***

```python
import numpy as np
import pandas as pd

data={
    "prize":[1000,1021,100000,1033,1044,1088,1100,100000],
    "discount":[20,23,25,27,55,28,29,50]
}
df=pd.DataFrame(data)
df["new"]=np.where (df["prize"] > 1033,True,False)
print(df)

```

# ***SIMPLE_IMPUTER***

```python
from sklearn.impute import SimpleImputer
import pandas as pd
import numpy as np

data={
    "marks":[99,98,98,99,98,99],
    "gender":["M","M","F","M","F","F"],
    "result":["good","excellent","super","super","hit","vvgood"]
}

df=pd.DataFrame(data)

a=SimpleImputer(missing_values=np.nan,strategy="mean")

result = df['marks'].values.reshape(-1, 1)

df["marks"] = a.fit_transform(result)
# b=a.fit_transform(df["marks"])
print(df)


```

```python
from sklearn.impute import SimpleImputer
import pandas as pd
import numpy as np

data={
    "marks":[99,98,98,np.nan,98,99],
    "gender":["M","M","F","M","F","F"],
    "result":["good","excellent","super","super","hit","vvgood"]
}

df=pd.DataFrame(data)

a=SimpleImputer(missing_values=np.nan,strategy="median")

result=df["marks"].values.reshape(-1,1)
df["marks"]=a.fit_transform(result)
print(df)
```
